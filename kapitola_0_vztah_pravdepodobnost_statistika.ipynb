# Úvod: Vztah mezi pravděpodobností a statistikou

Pravděpodobnost a statistika jsou dvě úzce propojené oblasti matematiky, které se zabývají nejistotou a informacemi získanými z dat. Přestože se často studují společně, jejich přístupy a cíle se liší.


## Pravděpodobnost jako matematický model

Pravděpodobnost je teoretický rámec, který popisuje chování náhodných jevů. Vychází z přesně definovaných množin, funkcí a pravidel. Pomocí pravděpodobnosti můžeme:

- modelovat náhodné pokusy (např. hod kostkou),
- definovat jevy a jejich pravděpodobnosti,
- odvozovat vlastnosti náhodných veličin.


Pravděpodobnostní model je tedy abstraktní a předpokládá, že známe zákonitosti, které řídí náhodný jev.

### Příklad: Hod kostkou

Množina možných výsledků:
$$ \Omega = \{1, 2, 3, 4, 5, 6\} $$

Pravděpodobnost, že padne sudé číslo:
$$ P(\text{sudé}) = \frac{3}{6} = 0{,}5 $$


```python
import random

omega = [1, 2, 3, 4, 5, 6]
jev_A = {2, 4, 6}

n = 10000
vysledky = [random.choice(omega) for _ in range(n)]
pocet_A = sum(1 for v in vysledky if v in jev_A)
P_A = pocet_A / n

print(f'Odhad pravděpodobnosti jevu A (sudé číslo): {P_A:.4f}')
```

## Statistika jako práce s daty

Statistika se zabývá sběrem, analýzou a interpretací dat. Na rozdíl od pravděpodobnosti pracuje s konkrétními pozorováními a snaží se z nich vyvodit závěry o populaci nebo procesu.

Statistika zahrnuje:

- popisná statistika (průměr, rozptyl, histogramy),
- statistická vizualizace, 
- statistická inference (odhady, testování hypotéz),
- regresní analýza, klasifikace, predikce.

Bez pravděpodobnostního modelu by inference nebyla možná – byla by to jen práce s čísly bez kontextu.

## Simulace jako propojení pravděpodobnosti a dat

Simulace představuje praktické propojení mezi pravděpodobnostním modelem a daty. Umožňuje nám generovat **umělá data** na základě známého pravděpodobnostního rozdělení. Tím můžeme ověřovat teoretické vlastnosti modelu, testovat algoritmy nebo ilustrovat chování náhodných jevů.

Simulace je užitečná například tehdy, když:
- nemáme k dispozici reálná data,
- chceme ověřit chování statistických metod,
- chceme vizualizovat náhodné jevy.

```python
import random
import matplotlib.pyplot as plt
# Simulace 10 000 hodů kostkou
n = 10000
omega = [1, 2, 3, 4, 5, 6]
vysledky = [random.choice(omega) for _ in range(n)]

# Histogram výsledků
plt.hist(vysledky, bins=range(1,8), align='left', edgecolor='black')
plt.title("Simulace hodu kostkou (n = 10 000)")
plt.xlabel("Výsledek")
plt.ylabel("Počet výskytů")
plt.grid(True)
plt.show()
```

```python
# Simulace 1000 hodnot z normálního rozdělení
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=0, scale=1, size=10)

# Histogram simulovaných dat
plt.hist(data, bins=30, edgecolor='black')
plt.title("Simulace dat z normálního rozdělení")
plt.xlabel("Hodnota")
plt.ylabel("Počet výskytů")
plt.grid(True)
plt.show()
```

## Statistická inference a role pravděpodobnosti

Statistická inference je proces, kdy z dat usuzujeme na vlastnosti celé populace. Abychom mohli dělat závěry, musíme předpokládat pravděpodobnostní model.

### Příklad: Odhad průměrné výšky

Předpokládáme, že výšky studentů mají normální rozdělení. Z výběru 30 studentů odhadneme průměrnou výšku celé populace.


```python
import numpy as np
import scipy.stats as stats

# Vzorek výšek
vysky = [170, 165, 180, 175, 160, 155, 185, 190, 165, 175]

mean = np.mean(vysky)
sem = stats.sem(vysky)
conf_int = stats.t.interval(0.95, len(vysky)-1, loc=mean, scale=sem)

print(f"Odhad průměrné výšky: {mean:.2f} cm")
print(f"95% interval spolehlivosti: ({conf_int[0]:.2f}, {conf_int[1]:.2f}) cm")
```
