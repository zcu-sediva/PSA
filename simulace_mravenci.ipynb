```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from ipywidgets import interact, IntSlider, FloatSlider

def spust_simulaci(pocet_mravencu=50, rychlost=1.0, delka_tyce=1.0):
    pozice = np.random.uniform(0, delka_tyce, pocet_mravencu)
    smer = np.random.choice([-1, 1], pocet_mravencu)
    aktivni = np.ones(pocet_mravencu, dtype=bool)

    fig, ax = plt.subplots()
    body, = ax.plot([], [], 'ro')
    ax.set_xlim(0, delka_tyce)
    ax.set_ylim(-0.1, 0.1)
    ax.set_title("Simulace pohybu mravenců na tyči")
    ax.set_xlabel("Pozice na tyči (m)")
    ax.set_yticks([])

    def update(frame):
        nonlocal pozice, smer, aktivni
        pozice[aktivni] += smer[aktivni] * rychlost * 0.02

        for i in range(pocet_mravencu):
            if aktivni[i] and (pozice[i] <= 0 or pozice[i] >= delka_tyce):
                aktivni[i] = False

        for i in range(pocet_mravencu):
            for j in range(i + 1, pocet_mravencu):
                if aktivni[i] and aktivni[j] and abs(pozice[i] - pozice[j]) < 0.01:
                    smer[i], smer[j] = -smer[i], -smer[j]

        body.set_data(pozice[aktivni], np.zeros(np.sum(aktivni)))
        return body,

    animace = FuncAnimation(fig, update, frames=300, interval=50, blit=True)
    plt.show()

interact(spust_simulaci,
         pocet_mravencu=IntSlider(min=10, max=100, step=1, value=50, description='Mravenci'),
         rychlost=FloatSlider(min=0.1, max=2.0, step=0.1, value=1.0, description='Rychlost (m/min)'),
         delka_tyce=FloatSlider(min=0.5, max=2.0, step=0.1, value=1.0, description='Délka tyče (m)'))
```
